# Diagnosis-Pair Subgroup Heatmap

Accuracy on five post-hoc diagnosis-pair subsets, including RoBERTa seed 20260718 and the across-subset mean.

This notebook regenerates the figure from the repository's saved evaluation outputs. It calls the canonical plotting function so notebook and manuscript figures cannot silently diverge. Running it exports editable SVG, vector PDF, and 600-DPI PNG files to `figures/`.


## Setup and project paths


```python
from pathlib import Path
import sys

from IPython.display import Image, display


def find_project_root() -> Path:
    """Find the repository root whether Jupyter starts here or above it."""
    start = Path.cwd().resolve()
    for candidate in (start, *start.parents):
        if (candidate / "analysis" / "publication_figures" / "generate_publication_figures.py").exists():
            return candidate
    raise FileNotFoundError("Could not locate analysis/publication_figures/generate_publication_figures.py")


ROOT = find_project_root()
if str(ROOT) not in sys.path:
    sys.path.insert(0, str(ROOT))

from analysis.publication_figures.generate_publication_figures import configure_style, contrast_heatmap

OUTPUT = ROOT / "figures" / "natural_contrast_heatmap.png"
print(f"Project root: {ROOT}")
print(f"Output: {OUTPUT}")
```

## Generate and preview


```python
configure_style()
contrast_heatmap()

assert OUTPUT.exists(), f"Figure was not created: {OUTPUT}"
display(Image(filename=str(OUTPUT)))
```

## Verify exported artifacts


```python
for suffix in ("svg", "pdf", "png"):
    artifact = OUTPUT.with_suffix(f".{suffix}")
    assert artifact.exists() and artifact.stat().st_size > 0, artifact
    print(f"{artifact.name}: {artifact.stat().st_size:,} bytes")
```
