# Cue-Masking Effects

Paired macro-F1 changes after formatting-marker removal and diagnosis-tail removal.

This notebook regenerates the figure from the repository's saved evaluation outputs. It calls the canonical plotting function so notebook and manuscript figures cannot silently diverge. Running it exports editable SVG, vector PDF, and 600-DPI PNG files to `figures/`.


## Setup and project paths


```python
from pathlib import Path
import sys

from IPython.display import Image, display


def find_project_root() -> Path:
    """Find the repository root whether Jupyter starts here or above it."""
    start = Path.cwd().resolve()
    for candidate in (start, *start.parents):
        if (candidate / "scripts" / "generate_publication_figures.py").exists():
            return candidate
    raise FileNotFoundError("Could not locate scripts/generate_publication_figures.py")


ROOT = find_project_root()
if str(ROOT) not in sys.path:
    sys.path.insert(0, str(ROOT))

from scripts.generate_publication_figures import configure_style, masking_effects

OUTPUT = ROOT / "figures" / "paper" / "cue_masking_effects.png"
print(f"Project root: {ROOT}")
print(f"Output: {OUTPUT}")
```

## Generate and preview


```python
configure_style()
masking_effects()

assert OUTPUT.exists(), f"Figure was not created: {OUTPUT}"
display(Image(filename=str(OUTPUT)))
```

## Verify exported artifacts


```python
for suffix in ("svg", "pdf", "png"):
    artifact = OUTPUT.with_suffix(f".{suffix}")
    assert artifact.exists() and artifact.stat().st_size > 0, artifact
    print(f"{artifact.name}: {artifact.stat().st_size:,} bytes")
```
